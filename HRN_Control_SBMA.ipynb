```python
import pandas as pd

df = pd.read_csv("SBMAsomascan_all.csv")

#後で使う、物質名リストcを取得

c = list(df.columns)

c.remove('Entrez gene symbol')
c.remove('responder/non-responder')
c.remove('group')

print(c)
```

```python
#H,N,R,さらにcontrolとsbmaのデータを、SBMAsomascan_all.csvから抽出する形で取得

df_h = df[(df["responder/non-responder"] == 0)]
df_n = df[(df["responder/non-responder"] == 1)]
df_r = df[(df["responder/non-responder"] == 2)]

df_control = df[(df["group"] == "Control")]
df_sbma = df[(df["group"] == "SBMA")]
```

```python
#t検定実行~ControlとSBMA

from scipy import stats

q_names = []
t_result = []
p_values_ttest = []
welch_result = []
p_values_welch = []

for question in c:
    t_statistic, p_value_ttest = stats.ttest_ind(df_control[question], df_sbma[question], equal_var=True)
    
    t_result.append(t_statistic)
    
    p_values_ttest.append(p_value_ttest)

    t_statistic_welch, p_value_welch = stats.ttest_ind(df_control[question], 
                                                   df_sbma[question], equal_var=False)
    
    welch_result.append(t_statistic_welch)
    p_values_welch.append(p_value_welch)
    
    q_names.append(question)

print(p_values_ttest)
```

```python
#ファイル出力
df_csv = pd.DataFrame({'proteins': q_names, 't_statistic': t_result, 'p_values_ttest':p_values_ttest,
                  'welch_result':welch_result , 'p_values_welch':p_values_welch})
df_csv.to_csv("pvalues_control_sbma.csv")
```

```python
#ファイル出力
df_csv = pd.DataFrame({'proteins': q_names, 't_statistic': t_result, 'p_values_ttest':p_values_ttest,
                  'welch_result':welch_result , 'p_values_welch':p_values_welch})
df_csv.to_csv("pvalues_HvsR.csv")
```

```python
#以下からFC導出過程

import numpy as np

#各群の平均値
m_con = df_control.mean()
m_sbma = df_sbma.mean()
m_h = df_h.mean()
m_r = df_r.mean()
m_n = df_n.mean()
```

```python
#上記で得られた各群の平均値から不要な列削除

m_n = m_n.drop('responder/non-responder')
m_r = m_r.drop('responder/non-responder')
m_h = m_h.drop('responder/non-responder')
m_con = m_con.drop('responder/non-responder')
m_sbma = m_sbma.drop('responder/non-responder')
```

```python
#log2FCの導出(ここではcon/sbma。同様にして、直上のセルの値を用いて導出。)

fcs = []

for i in range(len(m_con)):
    fc1 = m_con[i]/m_sbma[i]
    fc2 = np.log2(fc1)
    fcs.append(fc2)


print(fcs)
```

```python
#ファイル出力
df_csv = pd.DataFrame({'proteins': c, 'log2FC':fcs})
df_csv.to_csv("log2FC_CONperSBMA.csv")
```
