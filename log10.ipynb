```python
import pandas as pd
import numpy as np

def log10_transform_csv(input_file, output_file):
    # CSVファイルの読み込み
    df = pd.read_csv(input_file)
    
    # 数値のみに対して -log10 の変換を行う
    df_transformed = df.applymap(lambda x: -np.log10(x) if np.issubdtype(type(x), np.number) and x > 0 else x)
    
    # 結果を新しいCSVファイルに保存
    df_transformed.to_csv(output_file, index=False)
    print(f"Transformed CSV saved to: {output_file}")

# 使用例
input_file = 'q_pvalues_NvsH.csv'   # 変換元のCSVファイルのパス
output_file = "log10_" + input_file # 出力先のCSVファイルのパス
log10_transform_csv(input_file, output_file)
```
